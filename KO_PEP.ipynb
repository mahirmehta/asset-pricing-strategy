```python
import yfinance as yf
import pandas as pd
from lets_plot import *
LetsPlot.setup_html()
```

```python
tickers = ["KO", "PEP", "SPY"]
data = yf.download(
    tickers,
    start="2015-01-01",
    end="2026-01-01",
    auto_adjust=True
)

prices = data["Close"].dropna()
pair_prices = prices[["KO", "PEP"]]
returns = pair_prices.pct_change().dropna()
correlation = returns["KO"].corr(returns["PEP"])

print("Correlation: ", correlation)
```

```python
normalized_prices = pair_prices / pair_prices.iloc[0] * 100

normalized_plot = (
    normalized_prices
    .reset_index()
    .melt(
        id_vars="Date",
        var_name="Stock",
        value_name="Normalized Price"
    )
)

ggplot(
    normalized_plot,
    aes(x="Date", y="Normalized Price", color="Stock")
) + \
    geom_line() + \
    labs(
        title="Normalized Prices: KO vs PEP",
        x="Date",
        y="Index (Start = 100)"
    ) + \
    ggsize(800, 400)
```
